## Merging the dataframes

Core unit of analysis: **site x species x year**.

- `df_pheno` and `df_site_indices` both have site/species/year granularity, so we merge them on those keys.
- `df_site_location` is site-level metadata (one row per site), so we merge it in on site number.
- `df_collated_indices` is species/year-level (national index, no site), kept separate for species-level trend comparisons.

```python
print('df_pheno cols: ', df_pheno.columns)
print('df_site_indices cols: ', df_site_indices.columns)
print('df_site_location cols: ', df_site_location.columns)
print('df_species_trends cols: ', df_species_trends.columns)
print('df_collated_indices cols: ', df_collated_indices.columns)
```

```python
# standardise the join keys across dataframes before merging
df_pheno = df_pheno.rename(columns={'SITENO': 'SITE_CODE', 'SPECIES_NAME': 'SPECIES'})
df_site_location = df_site_location.rename(columns={'Site_Number': 'SITE_CODE'})

# 1. site x species x year: phenology + abundance index
df_site_species_year = pd.merge(
    df_pheno,
    df_site_indices,
    on=['SITE_CODE', 'SPECIES', 'YEAR'],
    how='inner',
    suffixes=('_pheno', '_idx'),
)

# 2. bring in site-level metadata (location, country, survey history)
df_merged = pd.merge(
    df_site_species_year,
    df_site_location,
    on='SITE_CODE',
    how='left',
    suffixes=('', '_site'),
)

print(df_pheno.shape, df_site_indices.shape, '->', df_site_species_year.shape)
print(df_site_species_year.shape, df_site_location.shape, '->', df_merged.shape)
df_merged.head()
```

## Exploratory Data Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

### 1. Survey coverage over time

How many sites and species are recorded each year — useful to know since UKBMS coverage grew substantially after 1976.

```python
coverage = df_merged.groupby('YEAR').agg(
    n_sites=('SITE_CODE', 'nunique'),
    n_species=('SPECIES', 'nunique'),
    n_records=('SPECIES', 'size'),
)

fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.plot(coverage.index, coverage['n_sites'], color='tab:blue', label='Sites surveyed')
ax1.set_ylabel('Number of sites', color='tab:blue')
ax1.set_xlabel('Year')

ax2 = ax1.twinx()
ax2.plot(coverage.index, coverage['n_species'], color='tab:orange', label='Species recorded')
ax2.set_ylabel('Number of species', color='tab:orange')

fig.suptitle('UKBMS survey coverage over time')
fig.tight_layout()
plt.show()
```

### 2. Phenology: mean flight date by species

Boxplot of `MEAN_FLIGHT_DATE` (day of year) across sites/years, for the 15 most-recorded species — shows which species fly early vs. late in the season and how variable their timing is.

```python
top_species = df_merged['COMMON_NAME_pheno'].value_counts().head(15).index

order = (
    df_merged[df_merged['COMMON_NAME_pheno'].isin(top_species)]
    .groupby('COMMON_NAME_pheno')['MEAN_FLIGHT_DATE']
    .median()
    .sort_values()
    .index
)

plt.figure(figsize=(10, 7))
sns.boxplot(
    data=df_merged[df_merged['COMMON_NAME_pheno'].isin(top_species)],
    x='MEAN_FLIGHT_DATE',
    y='COMMON_NAME_pheno',
    order=order,
)
plt.xlabel('Mean flight date (day of year)')
plt.ylabel('')
plt.title('Mean flight date by species (15 most-recorded species)')
plt.tight_layout()
plt.show()
```

### 3. Abundance vs. phenology: does flight period length relate to peak count?

Scatter of `FLIGHTPERIOD_RANGE` (days from first to last sighting) against `PEAKCOUNT`, log-scaled, to check whether species with a longer flight season also tend to be recorded in higher numbers at their peak.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_merged.sample(20000, random_state=42),
    x='FLIGHTPERIOD_RANGE',
    y='PEAKCOUNT',
    alpha=0.2,
    s=10,
)
plt.yscale('log')
plt.xlabel('Flight period range (days)')
plt.ylabel('Peak count (log scale)')
plt.title('Flight period length vs. peak count (20k row sample)')
plt.tight_layout()
plt.show()

df_merged[['FLIGHTPERIOD_RANGE', 'PEAKCOUNT', 'SITE_INDEX']].corr()
```

### 4. National abundance trends for a few species

Using `df_collated_indices` (species x year national collated index, England) to see long-term population trends for some well-known species.

```python
species_of_interest = ['Peacock', 'Small Tortoiseshell', 'Gatekeeper', 'Painted Lady', 'Marbled White']

trend_data = df_collated_indices[
    (df_collated_indices['COMMON_NAME'].isin(species_of_interest))
    & (df_collated_indices['COUNTRY'] == 'England')
]

plt.figure(figsize=(10, 6))
sns.lineplot(data=trend_data, x='YEAR', y='COLLATED_INDEX', hue='COMMON_NAME', marker='o', markersize=4)
plt.ylabel('Collated index (England)')
plt.title('National abundance trends, selected species')
plt.tight_layout()
plt.show()
```

### 5. Map of survey sites

Scatter of site Easting/Northing (British National Grid), coloured by country — a quick sanity check on the geographic spread of UKBMS sites.

```python
sites = df_site_location.dropna(subset=['Easting', 'Northing'])

plt.figure(figsize=(7, 9))
sns.scatterplot(data=sites, x='Easting', y='Northing', hue='Country', alpha=0.4, s=10, linewidth=0)
plt.gca().set_aspect('equal')
plt.title('UKBMS site locations')
plt.tight_layout()
plt.show()
```
